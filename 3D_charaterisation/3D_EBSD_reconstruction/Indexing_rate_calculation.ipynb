```python
import h5py
import numpy as np
```

## Calculate total indexing

```python
file_path = "E:/Users/lagon/The University of Manchester Dropbox/Laura Gonzalez/Zr-hydrides/Lunty polish/Lunty avec tickle/Laura analysis Part 1/3D microstructure analysis/"
file = file_path + "2_5_Recrystallised_rotated_aligned_cropped_indexingcalc.dream3d"

with h5py.File(file, 'r') as f:

    geom_path = "DataContainers/ImageDataContainer/_SIMPL_GEOMETRY/DIMENSIONS"
    dimensions = f[geom_path][()]
    
    # Extract the dimensions of the dataset:
    nx, ny, nz = dimensions
    total_voxels = nx * ny * nz
    
    print(f"Dimensions: {nx} x {ny} x {nz}")
    print(f"Total voxels: {total_voxels}")
    
    # Extract the phases:
    phases_path = "DataContainers/ImageDataContainer/CellData/Phases"
    phases = f[phases_path][()]
    
    # Extract the indexed voxels only:
    indexed_voxels = np.sum(phases > 0)

    print(f"Indexed voxels: {indexed_voxels}")
    
    # Calculate the indexing rate as indexed_voxels divided by the total number of voxels:
    indexing_rate = (indexed_voxels / total_voxels) * 100
    
    print(f"Indexing rate: {indexing_rate:.4f}%")
```

## Calculate indexing for Zr and hydride phases

```python
# Insert here the phase ID:
ZrID = 1
HydID = 2

zr_voxels  = int(np.count_nonzero(phases == ZrID))
hyd_voxels = int(np.count_nonzero(phases == HydID))


print("Zr indexing:", np.round(100*zr_voxels/total_voxels, 4),  "%")
print("Hydride indexing:", np.round(100*hyd_voxels/total_voxels, 4), "%")
```
